```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

corpus = '''
Hello, Welcome to My course, I am first second year student at SRM University in Chennai, I am very exited to explore the NLP course
'''

documents=sent_tokenize(corpus)

print(documents)

for sent in documents:
    print(sent)
```

```python
#tokenization
#paragraph into words
#sentence into words

from nltk.tokenize import word_tokenize

word_tokenize(corpus)

```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)

#in this fullistop is not considered as seperate word but at the end it is considered to be a seperate word
```

```python
###Stemming

words =["eating", "eats", "playing", "played", "plays", "play", "swimming", "swimms"]


###porter Stemmer

from nltk.stem import PorterStemmer

stemm = PorterStemmer()
print(stemm)

for word in words:
    print(word +"--->"+ stemm.stem(word))
```

```python
##RegexpStemmer class 

from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
# reg_stemmer = RegexpStemmer('ing|s|e$|able', min=4)

reg_stemmer.stem('ingeating')
```

```python

###SnowBall Stemmer

from nltk.stem import SnowballStemmer

snowballsstemmer = SnowballStemmer('english')

for word in words:
    print(word+ "--->" + snowballsstemmer.stem(word))
```

```python
##check what is the difference between porter stemming and snowball stemmer

stemm.stem("fairly") , stemm.stem("sportingly")

snowballsstemmer.stem("fairly"), snowballsstemmer.stem("sportingly")

#compare between these two snowball give right answer whereas porter stemming give wrong answer in some case 
```
