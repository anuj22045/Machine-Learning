```python
sentence  = "The Eiffel Tower was built from 1887 to 1889 by French Engineer Gustave Eiffel, whose company specialized in building mental frmaeworks and structures."
```

```python
import nltk
nltk.word_tokenize(sentence)
```

```python
words = nltk.word_tokenize(sentence)
tag_elements = nltk.pos_tag(words)
print(tag_elements)
```
